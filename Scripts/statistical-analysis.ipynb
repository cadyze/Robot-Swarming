```python
# Calculate the moments for the SR simulations
GRID_SIZE = 31
LOWER_BOUND = 0.018
UPPER_BOUND = 0.024
INCREMENT = 0.001


for lp in np.arange(LOWER_BOUND, UPPER_BOUND, INCREMENT):
    if(round(lp, 4) in LZ_PROB_MAP):
        (mean, var) = LZ_PROB_MAP[round(lp, 4)]
        print(f"LP: {round(lp, 4)} | MEAN & VAR: {round(mean, 2)} & {round(var, 2)}")
        continue
    mean, var = SimulatorMath.get_simulation_moments(GRID_SIZE, lp)
    LZ_PROB_MAP[round(lp, 4)] = (mean, var)
    print(f"LP: {round(lp, 4)} | MEAN: {round(mean, 2)} | VAR: {round(var, 2)}")
```
