```python
import pandas as pd

def calculate_prc(csv_path, sent):
    # Load CSV file
    df = pd.read_csv(csv_path)

    # Calculate total timesteps and total time waited
    total_timesteps = df['Timesteps'].sum()
    total_time_waited = df['Steps Waited'].sum()

    # Calculate percentage of time waited
    percentage_waited = (total_time_waited / total_timesteps) * 100

    print(f"Average Percentage of Time Waited{sent}: {percentage_waited:.2f}%")

print("A61 R5")
calculate_prc("A61-R5/A61_R5_WN_T4_FILL.csv", " at position T4")
calculate_prc("A61-R5/A61_R5_WN_T0_FILL.csv", " at position T0")

print("\nA61 R10")
calculate_prc("A61-R10/A61_R10_WN_T9_FILL.csv", " at position T9")
calculate_prc("A61-R10/A61_R10_WN_T0_FILL.csv", " at position T0")
```
